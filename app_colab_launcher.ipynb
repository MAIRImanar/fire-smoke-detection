```
# ── Étape 4 : Lancer l'app Streamlit avec ngrok ──────────────────
import subprocess, threading, time
from pyngrok import ngrok

# Lance Streamlit en arrière-plan
def run_streamlit():
    subprocess.run([
        'streamlit', 'run', 'app.py',
        '--server.port', '8501',
        '--server.headless', 'true',
        '--server.enableCORS', 'false',
        '--server.enableXsrfProtection', 'false',
    ])

t = threading.Thread(target=run_streamlit, daemon=True)
t.start()
time.sleep(5)  # Attendre que Streamlit démarre

# Créer un tunnel public avec ngrok
public_url = ngrok.connect(8501)
print('=' * 60)
print(f'  ✔ APPLICATION DISPONIBLE ICI :')
print(f'  {public_url}')
print('=' * 60)
print('  Clique sur le lien pour ouvrir l\'interface !')
```
